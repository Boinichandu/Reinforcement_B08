```python
# Discount factor
gamma = 0.9

# Policy
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

# State positions
states = {
    "S0": (0, 0),
    "S1": (0, 1),
    "G":  (0, 2),
    "S3": (1, 1),
    "S4": (1, 2)
}

# Rewards
normal_reward = -1
goal_reward = 10

# Find next state
def get_next_state(state, action):
    row, col = states[state]

    if action == "Up":
        row -= 1
    elif action == "Down":
        row += 1
    elif action == "Left":
        col -= 1
    elif action == "Right":
        col += 1

    # Check grid boundaries
    if row < 0 or row > 1 or col < 0 or col > 2:
        return state

    # Check obstacle
    if (row, col) == (1, 0):
        return state

    # Find corresponding state
    for s, position in states.items():
        if position == (row, col):
            return s

    return state


# Get reward
def get_reward(next_state):
    if next_state == "G":
        return goal_reward
    return normal_reward


# State Value Function
def state_value_function(state):
    total = 0
    discount = 1

    while state != "G":
        action = policy[state]

        next_state = get_next_state(state, action)

        reward = get_reward(next_state)

        total += discount * reward

        discount *= gamma

        state = next_state

    return total


# Action Value Function
def action_value_function(state, action):
    next_state = get_next_state(state, action)

    reward = get_reward(next_state)

    if next_state == "G":
        return reward

    return reward + gamma * state_value_function(next_state)


# Display State Value
print("State Value of S0:",
      state_value_function("S0"))

# Display Action Value
print("Action Value of S0 and Up:",
      action_value_function("S0", "Up"))
```
